```python
# Q3.2
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)
print(a is b)
```

```python
# Q3.3
x = 100
y = 100
print("x is y:", x is y)
print("x == y:", x == y)

p = 1000
q = 1000
print("p is q:", p is q)
print("p == q:", p == q)
p = int("1000")
q = int("1000")
print("p is q (built at run time):", p is q)
print("p == q:", p == q)
m = int("100")
n = int("100")
print("m is n (small ints):", m is n)          # True
```

# Topic 7: Functions & Passing Arguments

```python
# Q7.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val
print(my_abs(5))
print(my_abs(-5))
print(my_abs(0))
```

```python
# Q7.2
try:
    print(my_abs("Hi"))
except TypeError as e:
    print("TypeError:", e)
```

```python
# Q7.3
def print_abs(val):
    if val < 0:
        print(0 - val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)
```

```python
# Q7.4
def inc_val(val):
    val = val + 1

x = 7
inc_val(x)
print(x)
```

```python
# Q7.5
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print("Inside function after swap:", val1, val2)

x = 6
y = 3
swap_message(x, y)
print("Outside function (originals):", x, y)
```

# Topic 8: Scope — Local vs Global Variables

```python
# Q8.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val
    print(val)
print(my_abs(-4))
try:
    print(val)
except NameError as e:
    print("NameError:", e)
```

```python
# Q8.2
counter = 0

def show_counter():
    print(counter)

show_counter()
```

```python
# Q8.3
counter = 0

def increment_counter():
    counter = counter + 1

try:
    increment_counter()
except UnboundLocalError as e:
    print("UnboundLocalError:", e)
```

```python
# Q8.4
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
increment_counter()
print(counter)   # 2
```
