```python
# Q8.2
counter = 0

def show_counter():
    print(counter)

show_counter()
```

```python
# Q8.3
def increment_counter():
    counter = counter + 1

try:
    increment_counter()
except UnboundLocalError as e:
    print("Error:", e)
```

```python
# Q8.4
def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
