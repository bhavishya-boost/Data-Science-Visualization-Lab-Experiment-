```python
#35
rng = np.random.default_rng(42)
marks = rng.integers(20, 101, size=(12, 4))
student_avg = marks.mean(axis=1)
test_mean = marks.mean(axis=0)
test_std = marks.std(axis=0)
print("Student Averages:", student_avg)
print("Test Mean:", test_mean, "\nTest Std:", test_std)
print("Students passing all tests (>=40):", np.sum((marks >= 40).all(axis=1)))
print("Top student indices (avg >= 75):", np.where(student_avg >= 75)[0] + 1)
print("Test with highest standard deviation:", np.argmax(test_std) + 1)
```

```python
#36
m = np.arange(21, 46).reshape(5, 5)
central = m[1:4, 1:4]
product = central @ central.T
row_sums = product.sum(axis=1)
print("Row Sums:", row_sums)
print("Max Row Sum:", row_sums.max())
print("Indices with max sum:", np.where(row_sums == row_sums.max())[0])
```

```python
#37
rng = np.random.default_rng(42)
qty = rng.integers(10, 80, size=(6, 4))
unit_price = np.array([150, 240, 90, 320])
revenue = qty * unit_price
print("Revenue per Day:", revenue.sum(axis=1))
print("Revenue per Product:", revenue.sum(axis=0))
print("Day with highest revenue:", np.argmax(revenue.sum(axis=1)) + 1)
total_qty = qty.sum(axis=0)
print("Product with highest total quantity:", np.argmax(total_qty) + 1)
adjusted_price = np.where(total_qty > 100, unit_price * 0.9, unit_price)
print("Adjusted Total Revenue:", (qty * adjusted_price).sum())
```

```python
#38
rng = np.random.default_rng(42)
data = rng.normal(loc=50, scale=12, size=100)
print("Outliers count:", np.sum((data < 30) | (data > 70)))
cleaned = np.where(data < 30, 30, np.where(data > 70, 70, data))
print("Original stats (mean, var, std, min, max):", data.mean(), data.var(), data.std(), data.min(), data.max())
print("Cleaned stats (mean, var, std, min, max):", cleaned.mean(), cleaned.var(), cleaned.std(), cleaned.min(), cleaned.max())
```

```python
#39
branch_a = pd.Series({"Q1": 120, "Q2": 145, "Q4": 180})
branch_b = pd.Series({"Q2": 110, "Q3": 135, "Q4": 155})
branch_c = pd.Series({"Q1": 95, "Q3": 125, "Q4": 140})
combined = branch_a + branch_b + branch_c
print("Combined (sum of all with alignment):\n", combined)
common = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
print("Sum of common quarters:", combined[common].sum())
print("Combined (filled NaN with 0):\n", branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0))
```
