```python
#26
course_df = pd.DataFrame({'Student': ['A', 'B', 'C', 'D', 'E'],'Course': ['CSE', 'ECE', 'CSE', 'ME', 'ECE'],'Score': [88, 76, 90, 65, 82]})
print("Columns:", course_df.columns)
print("\nStudent Column:\n", course_df['Student'])
student_score_df = course_df[['Student', 'Score']]
print("\nStudent and Score DF:\n", student_score_df)
course_df['Batch'] = '2026'
print("\nUpdated course_df:\n", course_df)
```

```python
#27
travel_df = pd.DataFrame({'City': ['Delhi', 'Agra', 'Jaipur', 'Lucknow', 'Kanpur']})
travel_df['Code'] = np.arange(5)
print("After adding Code column:\n", travel_df)
code_series = pd.Series([101, 104], index=[1, 4])
travel_df['Code'] = code_series
print("\nAfter replacing Code column with indexed Series:\n", travel_df)
```

```python
#28
school_df = pd.DataFrame({'Student': ['Rohan', 'Soham', 'Reena'],'Class': [10, 10, 10],'Section': ['A', 'B', 'A']})
school_df2 = pd.DataFrame(school_df, columns=['Student', 'Class', 'Section', 'House'])
print("Before assigning House value:\n", school_df2)
school_df2['House'] = 'Red'
print("\nAfter assigning House value:\n", school_df2)
```

```python
#29
item_list = ['Pen', 'Pencil', 'Notebook', 'Eraser']
qty_list = [10, 20, 15, 30]
inventory_df = pd.DataFrame({'Item': item_list, 'Quantity': qty_list})
discount_series = pd.Series([5, 10], index=[0, 2])
inventory_df['Discount'] = discount_series
print("After adding Discount column:\n", inventory_df)
del inventory_df['Discount']
print("\nAfter deleting Discount column:\n", inventory_df)
```

```python
#30
lab_df = pd.read_clipboard()
print("1. Full DataFrame:\n", lab_df)
print("\n2. Column Names:\n", lab_df.columns)
single_word_col_name = [c for c in lab_df.columns if " " not in c][0]
print(f"\n3. Single-word column ({single_word_col_name}):\n", lab_df[single_word_col_name])
space_col_name = [c for c in lab_df.columns if " " in c][0]
print(f"\n4. Column with space ({space_col_name}):\n", lab_df[space_col_name])
small_df = lab_df[[single_word_col_name, space_col_name]]
print("\n5. Smaller DataFrame:\n", small_df)
lab_df['Status'] = 'Active'
print("\n6. DataFrame after adding Status column:\n", lab_df)
del lab_df['Status']
print("\n7. Final DataFrame after deleting Status column:\n", lab_df)
```
